# Top 7 Loss Functions to Evaluate Regression Models

This notebook introduces the most commonly used loss functions for evaluating regression models.

```
import numpy as np
from sklearn.metrics import mean_absolute_error, mean_squared_error, mean_absolute_percentage_error, r2_score
```

## Sample Data

```
y_true = np.array([100, 200, 300, 400, 500])
y_pred = np.array([110, 190, 290, 410, 480])
```

## 1. Mean Absolute Error (MAE)

```
mae = mean_absolute_error(y_true, y_pred)
print('MAE =', mae)
```

## 2. Mean Squared Error (MSE)

```
mse = mean_squared_error(y_true, y_pred)
print('MSE =', mse)
```

## 3. Root Mean Squared Error (RMSE)

```
rmse = np.sqrt(mean_squared_error(y_true, y_pred))
print('RMSE =', rmse)
```

## 4. Mean Absolute Percentage Error (MAPE)

```
mape = mean_absolute_percentage_error(y_true, y_pred)
print('MAPE =', mape)
```

## 5. Huber Loss

```
def huber_loss(y_true, y_pred, delta=1):
    error = y_true - y_pred
    is_small = np.abs(error) <= delta
    squared_loss = 0.5 * error**2
    linear_loss = delta * (np.abs(error) - 0.5 * delta)
    return np.where(is_small, squared_loss, linear_loss).mean()

print('Huber Loss =', huber_loss(y_true, y_pred))
```

## 6. Log-Cosh Loss

```
logcosh = np.mean(np.log(np.cosh(y_pred - y_true)))
print('Log-Cosh Loss =', logcosh)
```

## 7. R-Squared Score (R²)

```
r2 = r2_score(y_true, y_pred)
print('R² Score =', r2)
```
