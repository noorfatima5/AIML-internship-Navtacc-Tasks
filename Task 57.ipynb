## 2. Activation Functions

```
def sigmoid(x):
    return 1/(1+np.exp(-x))

def relu(x):
    return np.maximum(0, x)

def softmax(x):
    exp_x = np.exp(x - np.max(x))
    return exp_x / exp_x.sum(axis=0)
```

## 3. Forward Propagation Example

```
inputs = np.array([1.0, 2.0])
weights = np.array([[0.5, 0.3], [0.2, 0.8]])
bias = np.array([0.1, 0.2])

hidden = sigmoid(np.dot(inputs, weights) + bias)
print(hidden)
```

## 4. XOR Dataset

```
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

print(X)
print(y)
```

## 5. Simple MLP using Scikit-Learn

```
from sklearn.neural_network import MLPClassifier

model = MLPClassifier(hidden_layer_sizes=(4,), max_iter=1000, random_state=42)
model.fit(X, y.ravel())

predictions = model.predict(X)
print(predictions)
```
