# Understanding and Coding Neural Networks From Scratch in Python and R

This notebook introduces the fundamentals of neural networks and demonstrates how to implement them from scratch in both Python and R.

## 1. Import Libraries

```
# Python libraries
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Activation Functions

```
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def relu(x):
    return np.maximum(0, x)

print(sigmoid(0.5))
print(relu(-2))
```

## 3. Single Neuron Example

```
inputs = np.array([1, 2, 3])
weights = np.array([0.2, 0.8, -0.5])
bias = 2

output = sigmoid(np.dot(inputs, weights) + bias)
print('Neuron Output:', output)
```

## 4. Simple Feedforward Neural Network

```
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

np.random.seed(42)
weights = np.random.rand(2,1)
bias = np.random.rand(1)

for epoch in range(1000):
    output = sigmoid(np.dot(X, weights) + bias)

print(output)
```

## 5. R Implementation

```
# R Code
sigmoid <- function(x) {
  1 / (1 + exp(-x))
}

inputs <- c(1,2,3)
weights <- c(0.2,0.8,-0.5)
bias <- 2

output <- sigmoid(sum(inputs * weights) + bias)
print(output)
```
